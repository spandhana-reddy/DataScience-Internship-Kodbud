```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df.drop_duplicates()
```

```python
total_confirmed = df["Confirmed"].sum()

print("Total Confirmed Cases:", total_confirmed)
```

```python
total_deaths = df["Deaths"].sum()

print("Total Deaths:", total_deaths)
```

```python
total_recovered = df["Recovered"].sum()

print("Total Recovered:", total_recovered)
```

```python
total_active = df["Active"].sum()

print("Total Active Cases:", total_active)
```

```python
top_confirmed = df.nlargest(10, "Confirmed")

top_confirmed[["Country/Region", "Confirmed"]]
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    top_confirmed["Country/Region"],
    top_confirmed["Confirmed"]
)

plt.title("Top 10 Countries by Confirmed COVID-19 Cases")
plt.xlabel("Country")
plt.ylabel("Confirmed Cases")

plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../images/top_confirmed_cases.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
region_data = df.groupby("WHO Region")[[
    "Confirmed",
    "Deaths",
    "Recovered",
    "Active"
]].sum()

region_data
```

```python
plt.figure(figsize=(10, 6))

region_data["Confirmed"].sort_values().plot(kind="barh")

plt.title("Confirmed COVID-19 Cases by WHO Region")
plt.xlabel("Confirmed Cases")
plt.ylabel("WHO Region")

plt.tight_layout()
plt.show()
```

```python
highest_cases = df.loc[df["Confirmed"].idxmax()]

print("Country with highest confirmed cases:")
print(highest_cases[["Country/Region", "Confirmed"]])
```

```python
highest_deaths = df.loc[df["Deaths"].idxmax()]

print("Country with highest deaths:")
print(highest_deaths[["Country/Region", "Deaths"]])
```

```python
highest_recovered = df.loc[df["Recovered"].idxmax()]

print("Country with highest recoveries:")
print(highest_recovered[["Country/Region", "Recovered"]])
```

```python
print("Total Confirmed Cases:", df["Confirmed"].sum())
print("Total Deaths:", df["Deaths"].sum())
print("Total Recovered:", df["Recovered"].sum())
print("Total Active Cases:", df["Active"].sum())
```

```python
df["Death Rate"] = (df["Deaths"] / df["Confirmed"]) * 100
```

```python
top_death_rate = df.nlargest(10, "Death Rate")

top_death_rate[[
    "Country/Region",
    "Confirmed",
    "Deaths",
    "Death Rate"
]]
```

```python
df["Recovery Rate"] = (df["Recovered"] / df["Confirmed"]) * 100
```

```python
top_recovery_rate = df.nlargest(10, "Recovery Rate")

top_recovery_rate[[
    "Country/Region",
    "Confirmed",
    "Recovered",
    "Recovery Rate"
]]
```

```python
comparison = df.nlargest(10, "Confirmed")

comparison.set_index("Country/Region")[[
    "Confirmed",
    "Deaths",
    "Recovered"
]].plot(
    kind="bar",
    figsize=(14, 7)
)

plt.title("COVID-19 Cases, Deaths and Recoveries")
plt.xlabel("Country")
plt.ylabel("Number of Cases")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    "../images/covid_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
correlation = df[[
    "Confirmed",
    "Deaths",
    "Recovered",
    "Active",
    "New cases",
    "New deaths",
    "New recovered"
]].corr()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between COVID-19 Variables")
plt.tight_layout()

plt.savefig(
    "../images/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
results = df[[
    "Country/Region",
    "Confirmed",
    "Deaths",
    "Recovered",
    "Active",
    "New cases",
    "New deaths",
    "New recovered",
    "Death Rate",
    "Recovery Rate"
]]
```

```python
results.to_csv(
    "../results/covid_analysis_results.csv",
    index=False
)
```
