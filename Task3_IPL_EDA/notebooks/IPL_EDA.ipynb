```python
matches["winner"].value_counts()
```

```python
team_wins = matches["winner"].value_counts()

print(team_wins)
```

```python
top_10_winning_teams = team_wins.head(10)

print(top_10_winning_teams)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_10_winning_teams.values,
    y=top_10_winning_teams.index
)

plt.title("Top 10 IPL Teams by Number of Wins")
plt.xlabel("Number of Wins")
plt.ylabel("Team")

plt.tight_layout()

plt.savefig(
    "../images/top_10_winning_teams.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
winning_teams_df = team_wins.reset_index()

winning_teams_df.columns = ["Team", "Wins"]

winning_teams_df.to_csv(
    "../results/team_wins.csv",
    index=False
)

winning_teams_df.head(10)
```

```python
most_winning_team = team_wins.idxmax()
most_wins = team_wins.max()

print("Most winning team:", most_winning_team)
print("Number of wins:", most_wins)
```

```python
print(deliveries[["batsman", "batsman_runs"]].head())
```

```python
batsman_runs = deliveries.groupby("batsman")["batsman_runs"].sum()

print(batsman_runs.sort_values(ascending=False).head(10))
```

```python
top_10_scorers = batsman_runs.sort_values(
    ascending=False
).head(10)

print(top_10_scorers)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_10_scorers.values,
    y=top_10_scorers.index
)

plt.title("Top 10 IPL Run Scorers")
plt.xlabel("Total Runs")
plt.ylabel("Batsman")

plt.tight_layout()

plt.savefig(
    "../images/top_10_run_scorers.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
scorers_df = top_10_scorers.reset_index()

scorers_df.columns = ["Batsman", "Runs"]

scorers_df.to_csv(
    "../results/top_10_scorers.csv",
    index=False
)

scorers_df
```
