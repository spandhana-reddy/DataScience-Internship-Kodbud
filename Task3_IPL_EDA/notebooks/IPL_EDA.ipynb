```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
matches = pd.read_csv("../dataset/matches.csv")

matches.head()
```

```python

deliveries = pd.read_csv("../dataset/deliveries.csv/deliveries.csv")
```

```python

deliveries.head()
```

```python
print("Matches shape:", matches.shape)
print("Deliveries shape:", deliveries.shape)
```

```python
print("Matches columns:")
print(matches.columns.tolist())

print("\nDeliveries columns:")
print(deliveries.columns.tolist())
```

```python
matches["winner"].value_counts()
```

```python
team_wins = matches["winner"].value_counts()

print(team_wins)
```

```python
top_10_winning_teams = team_wins.head(10)

print(top_10_winning_teams)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_10_winning_teams.values,
    y=top_10_winning_teams.index
)

plt.title("Top 10 IPL Teams by Number of Wins")
plt.xlabel("Number of Wins")
plt.ylabel("Team")

plt.tight_layout()

plt.savefig(
    "../images/top_10_winning_teams.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
winning_teams_df = team_wins.reset_index()

winning_teams_df.columns = ["Team", "Wins"]

winning_teams_df.to_csv(
    "../results/team_wins.csv",
    index=False
)

winning_teams_df.head(10)
```

```python
most_winning_team = team_wins.idxmax()
most_wins = team_wins.max()

print("Most winning team:", most_winning_team)
print("Number of wins:", most_wins)
```

```python
print(deliveries[["batsman", "batsman_runs"]].head())
```

```python
batsman_runs = deliveries.groupby("batsman")["batsman_runs"].sum()

print(batsman_runs.sort_values(ascending=False).head(10))
```

```python
top_10_scorers = batsman_runs.sort_values(
    ascending=False
).head(10)

print(top_10_scorers)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_10_scorers.values,
    y=top_10_scorers.index
)

plt.title("Top 10 IPL Run Scorers")
plt.xlabel("Total Runs")
plt.ylabel("Batsman")

plt.tight_layout()

plt.savefig(
    "../images/top_10_run_scorers.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
scorers_df = top_10_scorers.reset_index()

scorers_df.columns = ["Batsman", "Runs"]

scorers_df.to_csv(
    "../results/top_10_scorers.csv",
    index=False
)

scorers_df
```

```python
top_scorer = batsman_runs.idxmax()
top_score = batsman_runs.max()

print("Top scorer:", top_scorer)
print("Total runs:", top_score)
```

```python
print(matches["venue"].head(10))
```

```python
venue_counts = matches["venue"].value_counts()

print(venue_counts)
```

```python
top_10_venues = venue_counts.head(10)

print(top_10_venues)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_10_venues.values,
    y=top_10_venues.index
)

plt.title("Top 10 IPL Venues by Number of Matches")
plt.xlabel("Number of Matches")
plt.ylabel("Venue")

plt.tight_layout()

plt.savefig(
    "../images/top_10_venues.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
venue_df = top_10_venues.reset_index()

venue_df.columns = ["Venue", "Matches"]

venue_df.to_csv(
    "../results/venue_trends.csv",
    index=False
)

venue_df
```

```python
most_used_venue = venue_counts.idxmax()
most_matches = venue_counts.max()

print("Most used venue:", most_used_venue)
print("Number of matches:", most_matches)
```

```python
toss_wins = matches["toss_winner"].value_counts()

print(toss_wins)
```

```python
top_10_toss_winners = toss_wins.head(10)

print(top_10_toss_winners)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_10_toss_winners.values,
    y=top_10_toss_winners.index
)

plt.title("Top 10 IPL Teams by Toss Wins")
plt.xlabel("Number of Toss Wins")
plt.ylabel("Team")

plt.tight_layout()

plt.savefig(
    "../images/top_10_toss_winners.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
toss_df = top_10_toss_winners.reset_index()

toss_df.columns = ["Team", "Toss_Wins"]

toss_df.to_csv(
    "../results/toss_winners.csv",
    index=False
)

toss_df
```

```python
toss_decisions = matches["toss_decision"].value_counts()

print(toss_decisions)
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(
    x=toss_decisions.index,
    y=toss_decisions.values
)

plt.title("IPL Toss Decisions")
plt.xlabel("Toss Decision")
plt.ylabel("Number of Matches")

plt.tight_layout()

plt.savefig(
    "../images/toss_decisions.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
toss_decision_df = toss_decisions.reset_index()

toss_decision_df.columns = ["Decision", "Count"]

toss_decision_df.to_csv(
    "../results/toss_decisions.csv",
    index=False
)
```

```python
matches["toss_and_match_winner"] = (
    matches["toss_winner"] == matches["winner"]
)
```

```python
toss_match_win_percentage = (
    matches["toss_and_match_winner"].mean() * 100
)

print(
    f"Teams winning the toss also won the match in "
    f"{toss_match_win_percentage:.2f}% of matches."
)
```

```python
matches.to_csv(
    "../results/matches_analyzed.csv",
    index=False
)
```

```python
summary = pd.DataFrame({
    "Analysis": [
        "Most Winning Team",
        "Most Wins",
        "Top Scorer",
        "Top Scorer Runs",
        "Most Used Venue",
        "Matches at Most Used Venue",
        "Toss Winner Also Won Match (%)"
    ],
    "Result": [
        most_winning_team,
        most_wins,
        top_scorer,
        top_score,
        most_used_venue,
        most_matches,
        round(toss_match_win_percentage, 2)
    ]
})

summary
```

```python
summary.to_csv(
    "../results/ipl_eda_summary.csv",
    index=False
)
```
